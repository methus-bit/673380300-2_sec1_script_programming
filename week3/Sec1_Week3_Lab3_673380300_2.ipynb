## Lab 3.1: Multiplication Table Generator

```python
def part1_single_table():
    print("=== Part 1: Single Number Table ===")
    try:
        num = int(input("Enter a number for the multiplication table (e.g., 7): "))
        print(f"\nMultiplication table for {num}:")
        for i in range(1, 13):
            print(f"{num} x {i} = {num * i}")
    except ValueError:
        print("Error: Please enter a valid integer.")

def part2_full_grid():
    print("\n=== Part 2: Full 12x12 Multiplication Grid ===")
    print("    ", end="")
    # Print column header
    for col in range(1, 13):
        print(f"{col:4d}", end="")
    print("\n" + "-" * 54)

    # Print rows
    for i in range(1, 13):
        print(f"{i:2d} |", end="")  # Row header
        for j in range(1, 13):
            print(f"{i*j:4d}", end="")
        print()  # New line after each row

if __name__ == "__main__":
    part1_single_table()
    part2_full_grid()
```

## Lab 3.2: Countdown & User Guessing Game

```python
def part1_countdown():
    print("=== Part 1: Countdown Timer ===")
    try:
        start = int(input("Enter a starting number for the countdown: "))
        count = start
        while count >= 0:
            print(count)
            count -= 1
        print("Blast off! 🚀")
    except ValueError:
        print("Error: Please enter a valid integer.")

def part2_guessing_game():
    print("\n=== Part 2: Simple Guessing Game ===")
    secret_number = 42
    max_attempts = 5
    attempts = 0

    print(f"I am thinking of a secret number. You have {max_attempts} attempts to guess it!")

    while attempts < max_attempts:
        try:
            attempts += 1
            guess = int(input(f"Guess the number (Attempt {attempts}/{max_attempts}): "))

            if guess == secret_number:
                print("Congratulations! You guessed it! 🎉")
                break
            elif guess < secret_number:
                print("Too low! Try again.")
            else:
                print("Too high! Try again.")
        except ValueError:
            print("Invalid input. Please enter an integer.")
    else:
        # ทำงานเฉพาะเมื่อลูปจบลงโดยไม่ถูก break (ทายไม่ถูกใน 5 ครั้ง)
        print(f"\nSorry, you ran out of attempts. The secret number was {secret_number}.")

if __name__ == "__main__":
    part1_countdown()
    part2_guessing_game()
```

## Activity 3.1: Debugging Infinite Loops & Loop Logic Errors

```python
def snippet_a_fixed():
    print("--- Buggy Snippet A (Fixed) ---")
    counter = 0
    while counter < 5:
        print(f"Counter is: {counter}")
        counter += 1  # เพิ่มการอัปเดตเพื่อหลีกเลี่ยง Loop ตัน

def snippet_b_fixed():
    print("\n--- Buggy Snippet B (Fixed) ---")
    print("Goal: Print 1 to 5")
    for i in range(1, 6):  # ปรับขอบเขตเป็น 6 เพื่อรันค่า 5 ด้วย
        print(i)

def snippet_c_fixed():
    print("\n--- Buggy Snippet C (Fixed) ---")
    for char in "hello":
        if char == 'l':
            print("Found 'l', stopping!")
            break
        print(char)

if __name__ == "__main__":
    snippet_a_fixed()
    snippet_b_fixed()
    snippet_c_fixed()
```

## Activity 3.2: Pattern Printing with Nested Loops

```python
def draw_square(size=5):
    print("--- 1. Square of Asterisks ---")
    for row in range(size):
        for col in range(size):
            print("* ", end="")
        print()

def draw_triangle(height=5):
    print("\n--- 2. Right-Angled Triangle ---")
    for row in range(1, height + 1):
        for col in range(row):
            print("* ", end="")
        print()

def draw_inverted_triangle(height=5):
    print("\n--- 3. Inverted Right-Angled Triangle ---")
    for row in range(height, 0, -1):
        for col in range(row):
            print("* ", end="")
        print()

def draw_pyramid(height=5):
    print("\n--- 4. Full Pyramid Challenge ---")
    for row in range(1, height + 1):
        # พิมพ์ช่องว่างนำหน้าเพื่อความสมดุล
        for space in range(height - row):
            print("  ", end="")
        # พิมพ์เครื่องหมายดาวตามสูตรเลขคี่ (2*n - 1)
        for col in range(2 * row - 1):
            print("* ", end="")
        print()

if __name__ == "__main__":
    draw_square()
    draw_triangle()
    draw_inverted_triangle()
    draw_pyramid()
```
