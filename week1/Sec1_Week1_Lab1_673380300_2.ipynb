2. Declare and print variables:

```python
age = 20  # integer
gpa = 3.75  # float
favorite_color = "Orange"  # string
is_student = True  # boolean

print("Age:", age,type(age))
print("GPA:", gpa,type(gpa))
print("Favourite Color:", favorite_color,type(favorite_color))
print("Student:", is_student,type(is_student))
```

3. Perform simple arithmetic:

```python
num1 = 6
num2 = 9

addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2
division = num1 / num2

print(f"addition: {num1} + {num2} = {addition}")
print(f"subtraction: {num1} - {num2} = {subtraction}")
print(f"multiplication: {num1} * {num2} = {multiplication}")
print(f"division: {num1} / {num2} = {division}")
```

การแก้โจทย์ปัญหาและปริศนา Python (Python Puzzlers / Solving Math Problems)

1. เขียนสคริปต์สั้น ๆ เพื่อแก้โจทย์ปัญหา (Word problems) เช่น การคำนวณพื้นที่สี่เหลี่ยม หรือการแปลงอุณหภูมิจากเซลเซียสเป็นฟาเรนไฮต์

2. Python Puzzlers: เป็นกิจกรรมเดี่ยวหรือคู่ที่ให้นักศึกษาดูตัวอย่างโค้ด (Snippets) แล้วทำตามขั้นตอน ดังนี้:คาดการณ์ผลลัพธ์ (Output): ทายว่าเมื่อรันโค้ดแล้วจะได้ผลลัพธ์อะไร

ระบุประเภทข้อมูล (Data Type): ระบุประเภทของตัวแปรหลังจากการดำเนินการต่าง ๆ

อธิบายเหตุผล: อธิบายว่าทำไมถึงคิดว่าได้ผลลัพธ์หรือประเภทข้อมูลเช่นนั้น เพื่อตรวจสอบความเข้าใจเรื่องการแปลงประเภทข้อมูล (Type Conversion) และความแตกต่างระหว่างข้อมูลประเภทต่าง ๆ เช่น Integer, Float, String และ Boolean


```python
# Puzzlers 2
name = "Alice"
age = "30"  # Note: This is a string
message = name + " is " + age + " years old."
print(message)
print(type(age))
```

ผลลัพธ์ (Output): `Alice is 30 years old.`

ประเภท DataType (ของตัวแปร age): `str` (String)

เหตุผล: ตัวแปร age ถูกกำหนดค่าให้อยู่ในเครื่องหมายคำพูด "30" จึงมีสถานะเป็นข้อความ (String) ไม่ใช่ตัวเลข และการใช้เครื่องหมาย + ระหว่างตัวแปรประเภท String ด้วยกัน เป็นการนำข้อความมาต่อกัน (Concatenation) ได้อย่างปกติ

```python
# Puzzlers 3
is_sunny = True
is_raining = False
print(is_sunny and is_raining)
print(type(is_sunny))
```

ผลลัพธ์ (Output): `False`

ประเภท DataType (ของตัวแปร is_sunny): `bool` (Boolean)

เหตุผล: ตัวแปร is_sunny เก็บค่าความจริงประเภทข้อมูล Boolean ส่วนผลลัพธ์ที่ได้ False เกิดจากการใช้ตัวดำเนินการตรรกศาสตร์ and ซึ่งเงื่อนไขจะเป็นจริง (True) ได้เพียงกรณีเดียวคือต้องเป็นจริงทั้งคู่ (True and True) เมื่อมีตัวใดตัวหนึ่งเป็นเท็จ ผลลัพธ์จึงออกมาเป็น False
